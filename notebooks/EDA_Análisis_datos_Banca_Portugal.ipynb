UNIÓN DE LOS DATOS DE CAMPAÑA Y CLIENTES -  mediante `id_` del CSV e `ID` del Excel - `Outer` conserva todos los registros para comprobar cuáles tienen coincidencia - `validate="1:1"` exige que los IDs sean únicos en ambos archivos.

CONVERTIR LAS CUATRO COLUMNAS NUMÉRICAS

```python

columnas_numericas = [
    "cons.price.idx", "cons.conf.idx",
    "euribor3m", "nr.employed"
]

for columna in columnas_numericas:
    original = Datos_limpios[columna].copy()

    convertido = pd.to_numeric(
        original.astype("string")
        .str.strip()
        .str.replace(",", ".", regex=False),
        errors="coerce"
    )

    errores = original.notna() & convertido.isna()

    print(f"{columna}: {errores.sum()} valores no convertibles")
    print(original.loc[errores].unique()[:10])

    Datos_limpios[columna] = convertido
```

COVERTIR FECHAS EN DATOS NUMÉRICOS

```python
meses = {
    "enero": "01", "febrero": "02", "marzo": "03",
    "abril": "04", "mayo": "05", "junio": "06",
    "julio": "07", "agosto": "08", "septiembre": "09",
    "setiembre": "09", "octubre": "10",
    "noviembre": "11", "diciembre": "12"
}

fecha_original = Datos_limpios["date"].copy()

fecha_texto = (
    fecha_original.astype("string")
    .str.strip()
    .str.lower()
)

fecha_normalizada = fecha_texto.str.replace(
    r"[a-záéíóúñ]+",
    lambda coincidencia: meses.get(
        coincidencia.group(0), coincidencia.group(0)
    ),
    regex=True
)

fecha_convertida = pd.to_datetime(
    fecha_normalizada,
    format="%d-%m-%Y",
    errors="coerce"
)

errores_fecha = fecha_original.notna() & fecha_convertida.isna()

print("Fechas no convertibles:", errores_fecha.sum())
print(fecha_original.loc[errores_fecha].unique()[:10])

Datos_limpios["date"] = fecha_convertida
```
